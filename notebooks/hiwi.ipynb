```python
import numpy as np
import open3d as o3d
import pye57 

# --- Step 1: Input E57 file path ---
# e57_path = '/Users/saidharb/Documents/LocalDocuments/HiWi - IGP/testfolder/PC2Model/test_output/furnished_house_12/scans/scan0' + scan_nr + '_points.e57'
# e57_path = '/Users/saidharb/Documents/LocalDocuments/HiWi - IGP/testfolder/PC2Model/demo_dataset_2/v1/point_clouds/mechanical_scans/mec_019-fusion360/scans/scan0' + scan_nr + '_points.e57'
e57_path = '/Users/saidharb/Documents/LocalDocuments/HiWi - IGP/testfolder/PC2Model/demo_dataset_2/v1/point_clouds/furniture_scans/fur_022-modelnet/scans/scan0' + scan_nr + '_points.e57'

# --- Step 2: Open the E57 file ---
e57 = pye57.E57(e57_path)

# Choose scan index (0 for first)
scan_index = 0

# Read the scan – request colors if present
data = e57.read_scan(scan_index, colors=False, ignore_missing_fields=True)
points = np.vstack((data["cartesianX"],
                    data["cartesianY"],
                    data["cartesianZ"])).T

print(f"Number of points: {points.shape[0]}")
pcd = o3d.geometry.PointCloud()
pcd.points = o3d.utility.Vector3dVector(points)

# --- Step 5: Save as PLY ---
# ply_path = '/Users/saidharb/Documents/LocalDocuments/HiWi - IGP/testfolder/PC2Model/images/house_12_scan_' + scan_nr + '.ply'
# ply_path = '/Users/saidharb/Documents/LocalDocuments/HiWi - IGP/testfolder/PC2Model/images/mech_019_' + scan_nr + '.ply'
ply_path = '/Users/saidharb/Documents/LocalDocuments/HiWi - IGP/testfolder/PC2Model/images/fur_022_' + scan_nr + '.ply'
o3d.io.write_point_cloud(ply_path, pcd)
print(f"PLY file saved at: {ply_path}")
```
