## CADTransformer

````python
from tqdm import tqdm
from dataset.cad_dataset import get_dataloader
from utils import ensure_dir
import torch
import numpy as np
import os
import h5py
from cadlib.macro import EOS_IDX
import torch.nn as nn
````

````python
class Bottleneck(nn.Module):
    def __init__(self, cfg):
        super(Bottleneck, self).__init__()

        self.bottleneck = nn.Sequential(nn.Linear(cfg.d_model, cfg.dim_z),
                                        nn.Tanh())

    def forward(self, z):
        return self.bottleneck(z)
````

````python
class FCN(nn.Module):
    def __init__(self, d_model, n_commands, n_args, args_dim=256):
        super().__init__()

        self.n_args = n_args
        self.args_dim = args_dim

        self.command_fcn = nn.Linear(d_model, n_commands)
        self.args_fcn = nn.Linear(d_model, n_args * args_dim)

    def forward(self, out):
        S, N, _ = out.shape

        command_logits = self.command_fcn(out)  # Shape [S, N, n_commands]

        args_logits = self.args_fcn(out)  # Shape [S, N, n_args * args_dim]
        args_logits = args_logits.reshape(S, N, self.n_args, self.args_dim)  # Shape [S, N, n_args, args_dim]

        return command_logits, args_logits
````

````python
class ConstEmbedding(nn.Module):
    """learned constant embedding"""
    def __init__(self, cfg, seq_len):
        super().__init__()

        self.d_model = cfg.d_model
        self.seq_len = seq_len

        self.PE = PositionalEncodingLUT(cfg.d_model, max_len=seq_len)

    def forward(self, z):
        N = z.size(1)
        src = self.PE(z.new_zeros(self.seq_len, N, self.d_model))
        return src
````

````python
class CADEmbedding(nn.Module):
    """Embedding: positional embed + command embed + parameter embed + group embed (optional)"""
    def __init__(self, cfg, seq_len, use_group=False, group_len=None):
        super().__init__()

        self.command_embed = nn.Embedding(cfg.n_commands, cfg.d_model)

        args_dim = cfg.args_dim + 1
        self.arg_embed = nn.Embedding(args_dim, 64, padding_idx=0)
        self.embed_fcn = nn.Linear(64 * cfg.n_args, cfg.d_model)

        # use_group: additional embedding for each sketch-extrusion pair
        self.use_group = use_group
        if use_group:
            if group_len is None:
                group_len = cfg.max_num_groups
            self.group_embed = nn.Embedding(group_len + 2, cfg.d_model)

        self.pos_encoding = PositionalEncodingLUT(cfg.d_model, max_len=seq_len+2)

    def forward(self, commands, args, groups=None):
        S, N = commands.shape

        src = self.command_embed(commands.long()) + \
              self.embed_fcn(self.arg_embed((args + 1).long()).view(S, N, -1))  # shift due to -1 PAD_VAL

        if self.use_group:
            src = src + self.group_embed(groups.long())

        src = self.pos_encoding(src)

        return src
````

````python
class Decoder(nn.Module):
    def __init__(self, cfg):
        super(Decoder, self).__init__()

        self.embedding = ConstEmbedding(cfg, cfg.max_total_len)

        decoder_layer = TransformerDecoderLayerGlobalImproved(cfg.d_model, cfg.dim_z, cfg.n_heads, cfg.dim_feedforward, cfg.dropout)
        decoder_norm = LayerNorm(cfg.d_model)
        self.decoder = TransformerDecoder(decoder_layer, cfg.n_layers_decode, decoder_norm)

        args_dim = cfg.args_dim + 1
        self.fcn = FCN(cfg.d_model, cfg.n_commands, cfg.n_args, args_dim)

    def forward(self, z):
        src = self.embedding(z)
        out = self.decoder(src, z, tgt_mask=None, tgt_key_padding_mask=None)

        command_logits, args_logits = self.fcn(out)

        out_logits = (command_logits, args_logits)
        return out_logits
````

````python
class Encoder(nn.Module):
    def __init__(self, cfg):
        super().__init__()

        seq_len = cfg.max_total_len
        self.use_group = cfg.use_group_emb
        self.embedding = CADEmbedding(cfg, seq_len, use_group=self.use_group)

        encoder_layer = TransformerEncoderLayerImproved(cfg.d_model, cfg.n_heads, cfg.dim_feedforward, cfg.dropout)
        encoder_norm = LayerNorm(cfg.d_model)
        self.encoder = TransformerEncoder(encoder_layer, cfg.n_layers, encoder_norm)

    def forward(self, commands, args):
        padding_mask, key_padding_mask = _get_padding_mask(commands, seq_dim=0), _get_key_padding_mask(commands, seq_dim=0)
        group_mask = _get_group_mask(commands, seq_dim=0) if self.use_group else None

        src = self.embedding(commands, args, group_mask)

        memory = self.encoder(src, mask=None, src_key_padding_mask=key_padding_mask)

        z = (memory * padding_mask).sum(dim=0, keepdim=True) / padding_mask.sum(dim=0, keepdim=True) # (1, N, dim_z)
        return z
````

````python
from model.layers.transformer import *
from model.layers.improved_transformer import *
from model.layers.positional_encoding import *
from model.model_utils import _make_seq_first, _make_batch_first, \
    _get_padding_mask, _get_key_padding_mask, _get_group_mask
class CADTransformer(nn.Module):
    def __init__(self, cfg):
        super(CADTransformer, self).__init__()

        self.args_dim = cfg.args_dim + 1

        self.encoder = Encoder(cfg)

        self.bottleneck = Bottleneck(cfg)

        self.decoder = Decoder(cfg)

    def forward(self, commands_enc, args_enc,
                z=None, return_tgt=True, encode_mode=False):
        commands_enc_, args_enc_ = _make_seq_first(commands_enc, args_enc)  # Possibly None, None

        if z is None:
            z = self.encoder(commands_enc_, args_enc_)
            z = self.bottleneck(z)
        else:
            z = _make_seq_first(z)

        if encode_mode: return _make_batch_first(z)

        out_logits = self.decoder(z)
        out_logits = _make_batch_first(*out_logits)

        res = {
            "command_logits": out_logits[0],
            "args_logits": out_logits[1]
        }

        if return_tgt:
            res["tgt_commands"] = commands_enc
            res["tgt_args"] = args_enc

        return res
````

## BaseTrainer

````python
import os
import torch
import torch.optim as optim
import torch.nn as nn
from abc import abstractmethod
from tensorboardX import SummaryWriter

class BaseTrainer(object):
    """Base trainer that provides common training behavior.
        All customized trainer should be subclass of this class.
    """
    def __init__(self, cfg):
        self.cfg = cfg

        self.log_dir = cfg.log_dir
        self.model_dir = cfg.model_dir
        self.clock = TrainClock()
        self.batch_size = cfg.batch_size

        # build network
        self.build_net(cfg)

        # set loss function
        self.set_loss_function()

        # set optimizer
        self.set_optimizer(cfg)

        # set tensorboard writer
        self.train_tb = SummaryWriter(os.path.join(self.log_dir, 'train.events'))
        self.val_tb = SummaryWriter(os.path.join(self.log_dir, 'val.events'))

    @abstractmethod
    def build_net(self, cfg):
        raise NotImplementedError

    def set_loss_function(self):
        """set loss function used in training"""
        pass

    def set_optimizer(self, cfg):
        """set optimizer and lr scheduler used in training"""
        self.optimizer = optim.Adam(self.net.parameters(), cfg.lr)
        self.scheduler = optim.lr_scheduler.StepLR(self.optimizer, cfg.lr_step_size)

    def save_ckpt(self, name=None):
        """save checkpoint during training for future restore"""
        if name is None:
            save_path = os.path.join(self.model_dir, "ckpt_epoch{}.pth".format(self.clock.epoch))
            print("Saving checkpoint epoch {}...".format(self.clock.epoch))
        else:
            save_path = os.path.join(self.model_dir, "{}.pth".format(name))

        if isinstance(self.net, nn.DataParallel):
            model_state_dict = self.net.module.cpu().state_dict()
        else:
            model_state_dict = self.net.cpu().state_dict()

        torch.save({
            'clock': self.clock.make_checkpoint(),
            'model_state_dict': model_state_dict,
            'optimizer_state_dict': self.optimizer.state_dict(),
            'scheduler_state_dict': self.scheduler.state_dict(),
        }, save_path)

        self.net#.cuda()

    def load_ckpt(self, name=None):
        """load checkpoint from saved checkpoint"""
        name = name if name == 'latest' else "ckpt_epoch{}".format(name)
        load_path = os.path.join(self.model_dir, "{}.pth".format(name))
        if not os.path.exists(load_path):
            raise ValueError("Checkpoint {} not exists.".format(load_path))

        checkpoint = torch.load(load_path, map_location=torch.device('cpu'), weights_only=True)
        print("Loading checkpoint from {} ...".format(load_path))
        if isinstance(self.net, nn.DataParallel):
            self.net.module.load_state_dict(checkpoint['model_state_dict'])
        else:
            self.net.load_state_dict(checkpoint['model_state_dict'])
        self.optimizer.load_state_dict(checkpoint['optimizer_state_dict'])
        self.scheduler.load_state_dict(checkpoint['scheduler_state_dict'])
        self.clock.restore_checkpoint(checkpoint['clock'])

    @abstractmethod
    def forward(self, data):
        """forward logic for your network"""
        """should return network outputs, losses(dict)"""
        raise NotImplementedError

    def update_network(self, loss_dict):
        """update network by back propagation"""
        loss = sum(loss_dict.values())
        self.optimizer.zero_grad()
        loss.backward()
        if self.cfg.grad_clip is not None:
            nn.utils.clip_grad_norm_(self.net.parameters(), self.cfg.grad_clip)
        self.optimizer.step()

    def update_learning_rate(self):
        """record and update learning rate"""
        self.train_tb.add_scalar('learning_rate', self.optimizer.param_groups[-1]['lr'], self.clock.epoch)
        self.scheduler.step()

    def record_losses(self, loss_dict, mode='train'):
        """record loss to tensorboard"""
        losses_values = {k: v.item() for k, v in loss_dict.items()}

        tb = self.train_tb if mode == 'train' else self.val_tb
        for k, v in losses_values.items():
            tb.add_scalar(k, v, self.clock.step)

    def train_func(self, data):
        """one step of training"""
        self.net.train()

        outputs, losses = self.forward(data)

        self.update_network(losses)
        if self.clock.step % 10 == 0:
            self.record_losses(losses, 'train')

        return outputs, losses

    def val_func(self, data):
        """one step of validation"""
        self.net.eval()

        with torch.no_grad():
            outputs, losses = self.forward(data)

        self.record_losses(losses, 'validation')

        return outputs, losses

    def visualize_batch(self, data, tb, **kwargs):
        """write visualization results to tensorboard writer"""
        raise NotImplementedError

class TrainClock(object):
    """ Clock object to track epoch and step during training
    """
    def __init__(self):
        self.epoch = 1
        self.minibatch = 0
        self.step = 0

    def tick(self):
        self.minibatch += 1
        self.step += 1

    def tock(self):
        self.epoch += 1
        self.minibatch = 0

    def make_checkpoint(self):
        return {
            'epoch': self.epoch,
            'minibatch': self.minibatch,
            'step': self.step
        }

    def restore_checkpoint(self, clock_dict):
        self.epoch = clock_dict['epoch']
        self.minibatch = clock_dict['minibatch']
        self.step = clock_dict['step']
````

## TrainerAE

````python
import torch
import torch.optim as optim
from tqdm import tqdm
from trainer.loss import CADLoss
from trainer.scheduler import GradualWarmupScheduler


class TrainerAE(BaseTrainer):
    def build_net(self, cfg):
        self.net = CADTransformer(cfg)#.cuda()

    def set_optimizer(self, cfg):
        """set optimizer and lr scheduler used in training"""
        self.optimizer = optim.Adam(self.net.parameters(), cfg.lr)
        self.scheduler = GradualWarmupScheduler(self.optimizer, 1.0, cfg.warmup_step)

    def set_loss_function(self):
        self.loss_func = CADLoss(self.cfg)#.cuda()

    def forward(self, data):
        commands = data['command']#.cuda() # (N, S)
        args = data['args']#.cuda()  # (N, S, N_ARGS)

        outputs = self.net(commands, args)
        loss_dict = self.loss_func(outputs)

        return outputs, loss_dict

    def encode(self, data, is_batch=False):
        """encode into latent vectors"""
        commands = data['command']#.cuda()
        args = data['args']#.cuda()
        if not is_batch:
            commands = commands.unsqueeze(0)
            args = args.unsqueeze(0)
        z = self.net(commands, args, encode_mode=True)
        return z

    def decode(self, z):
        """decode given latent vectors"""
        outputs = self.net(None, None, z=z, return_tgt=False)
        return outputs

    def logits2vec(self, outputs, refill_pad=True, to_numpy=True):
        """network outputs (logits) to final CAD vector"""
        out_command = torch.argmax(torch.softmax(outputs['command_logits'], dim=-1), dim=-1)  # (N, S)
        out_args = torch.argmax(torch.softmax(outputs['args_logits'], dim=-1), dim=-1) - 1  # (N, S, N_ARGS)
        if refill_pad: # fill all unused element to -1
            mask = ~torch.tensor(CMD_ARGS_MASK).bool()[out_command.long()] #.cuda() infront of [out_command.long()]
            out_args[mask] = -1

        out_cad_vec = torch.cat([out_command.unsqueeze(-1), out_args], dim=-1)
        if to_numpy:
            out_cad_vec = out_cad_vec.detach().cpu().numpy()
        return out_cad_vec

    def evaluate(self, test_loader):
        """evaluatinon during training"""
        self.net.eval()
        pbar = tqdm(test_loader)
        pbar.set_description("EVALUATE[{}]".format(self.clock.epoch))

        all_ext_args_comp = []
        all_line_args_comp = []
        all_arc_args_comp = []
        all_circle_args_comp = []

        for i, data in enumerate(pbar):
            with torch.no_grad():
                commands = data['command']#.cuda()
                args = data['args']#.cuda()
                outputs = self.net(commands, args)
                out_args = torch.argmax(torch.softmax(outputs['args_logits'], dim=-1), dim=-1) - 1
                out_args = out_args.long().detach().cpu().numpy()  # (N, S, n_args)

            gt_commands = commands.squeeze(1).long().detach().cpu().numpy() # (N, S)
            gt_args = args.squeeze(1).long().detach().cpu().numpy() # (N, S, n_args)

            ext_pos = np.where(gt_commands == EXT_IDX)
            line_pos = np.where(gt_commands == LINE_IDX)
            arc_pos = np.where(gt_commands == ARC_IDX)
            circle_pos = np.where(gt_commands == CIRCLE_IDX)

            args_comp = (gt_args == out_args).astype(np.int)
            all_ext_args_comp.append(args_comp[ext_pos][:, -N_ARGS_EXT:])
            all_line_args_comp.append(args_comp[line_pos][:, :2])
            all_arc_args_comp.append(args_comp[arc_pos][:, :4])
            all_circle_args_comp.append(args_comp[circle_pos][:, [0, 1, 4]])

        all_ext_args_comp = np.concatenate(all_ext_args_comp, axis=0)
        sket_plane_acc = np.mean(all_ext_args_comp[:, :N_ARGS_PLANE])
        sket_trans_acc = np.mean(all_ext_args_comp[:, N_ARGS_PLANE:N_ARGS_PLANE+N_ARGS_TRANS])
        extent_one_acc = np.mean(all_ext_args_comp[:, -N_ARGS_EXT_PARAM])
        line_acc = np.mean(np.concatenate(all_line_args_comp, axis=0))
        arc_acc = np.mean(np.concatenate(all_arc_args_comp, axis=0))
        circle_acc = np.mean(np.concatenate(all_circle_args_comp, axis=0))

        self.val_tb.add_scalars("args_acc",
                                {"line": line_acc, "arc": arc_acc, "circle": circle_acc,
                                 "plane": sket_plane_acc, "trans": sket_trans_acc, "extent": extent_one_acc},
                                global_step=self.clock.epoch)
````

````python
from torch.utils.data import Dataset, DataLoader
import torch
import os
import json
import h5py
import random

def get_dataloader(phase, config, shuffle=None):
    is_shuffle = phase == 'train' if shuffle is None else shuffle
    dataset = CADDataset(phase, config)
    dataloader = DataLoader(dataset, batch_size=config.batch_size, shuffle=is_shuffle, num_workers=config.num_workers,
                            worker_init_fn=np.random.seed())
    return dataloader


class CADDataset(Dataset):
    def __init__(self, phase, config):
        super(CADDataset, self).__init__()
        self.raw_data = os.path.join(config.data_root, "cad_vec") # h5 data root
        self.phase = phase
        self.aug = config.augment
        self.path = os.path.join(config.data_root, "train_val_test_split.json")
        with open(self.path, "r") as fp:
            self.all_data = json.load(fp)[phase]
        self.max_n_loops = config.max_n_loops          # Number of paths (N_P)
        self.max_n_curves = config.max_n_curves            # Number of commands (N_C)
        self.max_total_len = config.max_total_len
        self.size = 256

    def get_data_by_id(self, data_id):
        idx = self.all_data.index(data_id)
        return self.__getitem__(idx)

    def __getitem__(self, index):
        data_id = self.all_data[index]
        h5_path = os.path.join(self.raw_data, data_id + ".h5")
        with h5py.File(h5_path, "r") as fp:
            cad_vec = fp["vec"][:] # (len, 1 + N_ARGS)

        if self.aug and self.phase == "train":
            command1 = cad_vec[:, 0]
            ext_indices1 = np.where(command1 == EXT_IDX)[0]
            # if len(ext_indices1) > 1 and random.randint(0, 1) == 1:
            if len(ext_indices1) > 1 and random.uniform(0, 1) > 0.5:
                ext_vec1 = np.split(cad_vec, ext_indices1 + 1, axis=0)[:-1]
        
                data_id2 = self.all_data[random.randint(0, len(self.all_data) - 1)]
                h5_path2 = os.path.join(self.raw_data, data_id2 + ".h5")
                with h5py.File(h5_path2, "r") as fp:
                    cad_vec2 = fp["vec"][:]
                command2 = cad_vec2[:, 0]
                ext_indices2 = np.where(command2 == EXT_IDX)[0]
                ext_vec2 = np.split(cad_vec2, ext_indices2 + 1, axis=0)[:-1]
        
                n_replace = random.randint(1, min(len(ext_vec1) - 1, len(ext_vec2)))
                old_idx = sorted(random.sample(list(range(len(ext_vec1))), n_replace))
                new_idx = sorted(random.sample(list(range(len(ext_vec2))), n_replace))
                for i in range(len(old_idx)):
                    ext_vec1[old_idx[i]] = ext_vec2[new_idx[i]]
        
                sum_len = 0
                new_vec = []
                for i in range(len(ext_vec1)):
                    sum_len += len(ext_vec1[i])
                    if sum_len > self.max_total_len:
                        break
                    new_vec.append(ext_vec1[i])
                cad_vec = np.concatenate(new_vec, axis=0)

        pad_len = self.max_total_len - cad_vec.shape[0]
        cad_vec = np.concatenate([cad_vec, EOS_VEC[np.newaxis].repeat(pad_len, axis=0)], axis=0)

        command = cad_vec[:, 0]
        args = cad_vec[:, 1:]
        command = torch.tensor(command, dtype=torch.long)
        args = torch.tensor(args, dtype=torch.long)
        return {"command": command, "args": args, "id": data_id}

    def __len__(self):
        return len(self.all_data)
````

## CADTransformer
The input data is first fed to the CADTransformer which consists of a **Encoder**, **Bottleneck** and **Decoder**. 
The CADTransformer first permutes the order of dimensions from ```batch```x```commands```x```args``` to: ```commands```x```batch```x```args```

````python
commands_enc, args_enc = _make_seq_first(commands, args)  # Possibly None, None
print(commands_enc.shape)
print(args_enc.shape)
````

The forward method of the CADTransformer has different modes of operation: Either a latent vector **z** is given or not. If it is not given, which is the case we are looking at now, the commands and args are fed into the Encoder. The other case will be interesting as soon as we train a point cloud encoder for the same latent space **z**.

### Encoder

- ```seq_len``` is the maximum sequence length of commands, which is set to 60 according to the paper
- ```use_group``` is a boolean for to enable group learning

The Encoder first creates a ```padding_mask```, ```key_padding_mask``` and a ```group_mask```. The scripts are located in ```model_utils.py```.
- ```padding_mask``` creates a boolean int tensor which marks every command which is valid (in our case 6 commands are valid and the remaining 54 are not.
- ```key_padding_mask``` creates a boolean tensor marking all valid (i.e. all commands except EOS) commands with False. Later all comands marked with True (i.e. non-valid commands) are filled in with ```-inf``` during multi-head-attention.
- ```group_mask``` creates a tensor of length 60 (seq_length), where all commands belonging to an extrusion (cmd_idx: 5) sequence get the same number, i.e. [4, 2, 5, 4, 2, 5, 4, 0] --> [0, 0, 1, 1, 1, 2, 2, 2]

````python
# Encoder

## Definition
seq_len = cfg.max_total_len
print("Sequence length:", seq_len)
use_group = cfg.use_group_emb
print("Use group:", use_group)

## Forward
padding_mask, key_padding_mask = _get_padding_mask(commands_enc, seq_dim=0), _get_key_padding_mask(commands_enc, seq_dim=0)
group_mask = _get_group_mask(commands_enc, seq_dim=0) if use_group else None
````

After creating the masks, the Encoder embedds the command sequene using ```CADEmbedding```

#### CADEmbedding

The CADEmbedding module consists of lookup tables for the commands and args, as well as linear layers for transformation.
- ```command_embed``` (6, 256) -> Lookup table for each of the 6 commands with embedding of dim 256
- ```args_dim``` Adds another dimension to the parameters if the parameter is unused
- ```arg_embed``` (257, 64) -> Every parameter is quantized in $2^8$ bits and each of these $256$ bits gets an embedding of dim $64$ through a lookup table. The 257'th dimension is there if a parameter is not used in the command. All args with value 0 are not contributing to the gradient.
- ```embed_fcn``` (64 * 16 = 1024, 256) -> linear layer to project the flattened parameter embeddings to the model dimension
- ```group_embed``` (32,256) -> Lookup table for groups (max. 30 groups), 2 extra for beginning and end. There is no mention of that in the paper
- ```pos_encoding``` (64+2=66, 256) -> Learnable positional encoding for each of the 64 position, 2 extra for SOS and EOS

In the forward method the command and parameter embedding are added, as well as the positional encoding and the group embedding.

````python
# Embedding

## Defintion
command_embed = nn.Embedding(cfg.n_commands, cfg.d_model) 
args_dim = cfg.args_dim + 1 # 257 
arg_embed = nn.Embedding(args_dim, 64, padding_idx=0) 
embed_fcn = nn.Linear(64 * cfg.n_args, cfg.d_model) # 64 * 16 = 1024, 256 -> Projects the flattened args to dim 256
if use_group:
    group_len = cfg.max_num_groups # 30
    group_embed = nn.Embedding(group_len + 2, cfg.d_model) # whatever this group thing is, but it projects them from dim 32 to dim 256
pos_encoding = PositionalEncodingLUT(cfg.d_model, max_len=seq_len+2) # -> Adds a learnable positional encoding (initialized with kaiming)

## Forward
S, N = commands_enc.shape
src = command_embed(commands_enc.long())  
src += embed_fcn(arg_embed((args_enc + 1).long()).view(S, N, -1)) # shift due to -1 PAD_VAL
if use_group:
    src = src + group_embed(group_mask.long())
src = pos_encoding(src)
print(src.shape)
````

### Encoder (Continuation)

After the Embedding layer was called in the Encoder, the Encoder calls the encoder layer with this line on the embeddings: 

```memory = self.encoder(src, mask=None, src_key_padding_mask=key_padding_mask)```

The Encoder consists of multiple TransformerEncoderLayer's which are concatenetead in a list using ```_get_clones()```. Each TransformerEncoderLayer[Improved] is then called sequentially on the ```src``` values from the embedding. TransformerEncoderLayer[Improved] is basically a Transformer encoder with normalization --> self attention --> dropout --> residual connection --> normalization --> linear layer --> dropout --> residual connection. It would be too much and unnecessary to write all of this down, which is why the class is imported. **Important** to note is that the keys queries and values are calculated from the embeddings as usual. 

The embeddings of the commands are processed by ```self.n_layers = 4``` encoder layers and the attention mechanism will assign the attention. A final single layer of Layer normalization is applied afterwards (LN is already applied in the encoder).

To calculate the latent vector **z**, first the padding mask is applied to memory which leaves only memory entries at valid commands (all comands except EOS). Then the embeddings of all valid commands are averaged, i.e. they are added along the command dimension and divided by the number of commands (global averaging).

````python
# Encoder 2
def _get_clones(module, N):
    return ModuleList([copy.deepcopy(module) for i in range(N)])
    
encoder_layer = TransformerEncoderLayerImproved(cfg.d_model, cfg.n_heads, cfg.dim_feedforward, cfg.dropout)
layers = _get_clones(encoder_layer, cfg.n_layers)
encoder_norm = LayerNorm(cfg.d_model)

output = src
for mod in layers: # Encoder layers
    output = mod(output, memory2=None, src_mask=None, src_key_padding_mask=key_padding_mask)

if encoder_norm is not None:
    memory = encoder_norm(output)
print(memory.shape)

z = (memory * padding_mask).sum(dim = 0, keepdim = True)/padding_mask.sum(dim = 0, keepdim = True)
print(z.shape)
````

## Bottleneck

After the Encoder encoded the command and parameter sequence, the latent vector **z** is passed through the bottleneck. The bottleneck consists of a simple linear layer with a tanh non-linearity.

````python
bottleneck = nn.Sequential(nn.Linear(cfg.d_model, cfg.dim_z), nn.Tanh())
z = bottleneck(z)
print(z.shape)
````

## Decoder

After the bottleneck, the Decoder will try to decode the latent vector back to the original command sequence. The latent vector is first embedded and then decoded by the Decoder. Finally the decoded latent vector is passed through a Fully Connected Layer which results in the command and parameter logits. We will go step by step and start with the ```ConstEmbedding``` which is used for embedding the latent vector.

### Embedding
The embedding consist only of a positional encoding using a learnable lookup table (LUT) which will be shown now.

````python
N = z.size(1)
````

#### Learnable Positional Encoding

For the positional encoding, the dropout layer, a tensor consisiting of all the 60 (seq_length) positions (0 to 59) and an embedding/lookup table is initialized. Kaiming weight initalization is applied to the lookup table. In the forward method, the following happens: 
- ```z_new``` is initialized with the ```new_zeros()``` method, which creates a zero tensor of the same shape and the same device as the argument tensor --> This is basically to prevent compatibility issues
- ```pos``` checks the valid positions of the tensor, i.e. if the tensors sequence lenght is 60, only 60 positions are valid

The positional embedding is then added on top of the newly created zero tensor and the positional encoding is obtained. The positional encoding will the be passed alongside the latent vector **z** to the decoder.

````python
## Definition
dropout = nn.Dropout(p = 0.1)
position = torch.arange(0, seq_len, dtype = torch.long).unsqueeze(1)
pos_embed = nn.Embedding(60, 256)
nn.init.kaiming_normal_(pos_embed.weight, mode="fan_in")

## Forward
z_new = z.new_zeros(60, 1, 256)
pos = position[:z_new.size(0)]
z_new = z_new + pos_embed(pos)
src = dropout(z_new)
````

## Decoder (Continuation)

After the positional encoding was obtained, it is passed alongside the latent vector **z** to the decoder layers. The decoder first copies the decoder layers. The input to a decoder layer is the positional encoding ```src``` and the latent vector **z**. A decoder layer is basically the decoder layer from the Attention is all you need paper. It takes the positional encoding of the input sequence as input to the first decoder block, i.e. the positional encoding is normalized, self attention is applied and a dropout layer. The original input is then added in a residual connection. In the second decoder block cross-attention is applied, i.e. the self attention values from the positional encodings are normalized and used as queries and the latent vector **z** is used as key and value values. This basically means, that the decoder looks into the keys of the encoder by using the queries to know how to further decode the sequence. The answer to this comparison (dot product similarity) is reflected in the attention values. Again a residual connection with droput combines the outputs of the frist and second decoder block. A final normlization, linear and dropout layer is used to generate the outputs. After this a final normalization layer is applied.

````python
n_layers_dedecoder_layer = TransformerDecoderLayerGlobalImproved(cfg.d_model, cfg.dim_z, cfg.n_heads, cfg.dim_feedforward, cfg.dropout)
layers = _get_clones(decoder_layer, 4)
decoder_norm = LayerNorm(cfg.d_model)

for mod in layers:
    output = mod(src, # positional embedding
                 z) # latent vector z
if decoder_norm is not None:
    output = decoder_norm(output)
````

### Fully Connected Layer

After the Decoder used the attention mechanism to refine the outputs of the encoder and combined them with the sequence, a Fully Connected Layer is employed to get the final command and parameter logits. The FCN consists of two simple linear layers, one for the commands and one for the parameters.

````python
## Definition
command_fcn = nn.Linear(256, 6)
args_fcn = nn.Linear(256, 16 * 256) #n_args * args_dim  rfub fvub

## Forward
command_logits = command_fcn(output)  # Shape [S, N, n_commands]
args_logits = args_fcn(output) # Shape [S, N, n_args * args_dim]
args_logits = args_logits.reshape(S, N, 16, 256)
````

````python
# Some geometry functions regarding the Euler angles
def rotate_by_y(vec, theta):
    mat = np.array([[np.cos(theta), 0, np.sin(theta)],
                    [0, 1, 0],
                    [-np.sin(theta), 0, np.cos(theta)]])
    return np.dot(mat, vec)


def rotate_by_z(vec, phi):
    mat = np.array([[np.cos(phi), -np.sin(phi), 0],
                    [np.sin(phi), np.cos(phi), 0],
                    [0, 0, 1]])
    return np.dot(mat, vec)
    
def polar2cartesian(vec):
    """convert a vector in polar(spherical) coordinates to cartesian coordinates"""
    r = 1 if len(vec) == 2 else vec[2]
    theta, phi = vec[0], vec[1]
    x = r * np.sin(theta) * np.cos(phi)
    y = r * np.sin(theta) * np.sin(phi)
    z = r * np.cos(theta)
    return np.array([x, y, z])
````

## **1. The GAN’s Adversarial Training Process**
The generator \( G \) and discriminator \( D \) are trained **simultaneously** in an adversarial manner. Here’s how it works:

1. **Generator's Role**:
   - The generator G takes random noise and outputs a latent vector z
   - At the beginning of training, the generator produces completely random outputs because its weights are randomly initialized.

2. **Discriminator's Role**:
   - The discriminator D takes a latent vector z (either real z_real or fake z_fake) and outputs a score D(z):
     - \( D(z) \) is **high** for real latent vectors z_real (from the autoencoder).
     - \( D(z) \) is **low** for fake latent vectors z_gen (from the generator).

3. **Adversarial Objective**:
   - The discriminator \( D \) tries to **maximize** the difference between real and fake scores:
   - The generator \( G \) tries to **fool** the discriminator by producing fake latent vectors z_gen that look like real ones. The generator’s loss is:
     - This means the generator tries to **minimize** the discriminator’s score for fake latent vectors, essentially tricking the discriminator into believing z_gen are real.

---

## **2. How the Generator Learns to Fool the Discriminator**
Even though the generator starts with random noise, it gradually learns to produce realistic latent vectors through **gradient-based optimization**.

Here’s how:

1. **Discriminator Backpropagation**:
   - During training, the discriminator \( D \) is updated to distinguish real latent vectors \( z_{\text{real}} \) from fake latent vectors \( z_{\text{gen}} \).
   - The discriminator’s gradients flow backward to improve its ability to classify real and fake vectors.

2. **Generator Backpropagation**:
   - The generator \( G \) does not directly see the real latent vectors. Instead, it **receives feedback** from the discriminator.
   - Specifically, the gradients of the discriminator’s output \( D(z_{\text{gen}}) \) are backpropagated **through the generator** to update its weights.
   - Intuitively:
     - If D(z_gen) is low (discriminator says "fake"), the generator gets a **strong gradient signal** to adjust its weights.
     - The generator updates its parameters so that the next batch of z_gen is closer to the real latent vectors z_real.

3. **Iterative Improvement**:
   - Over many iterations, the generator \( G \) learns to produce latent vectors z_gen that are indistinguishable from z_real because:
     - The discriminator keeps improving its ability to tell real from fake.
     - The generator keeps improving its outputs to fool the discriminator.

---

## **3. The Role of Noise Input**
The random noise serves as a **seed** for the generator to produce diverse outputs. Without noise:
- The generator might always produce the same output (collapse issue).
- Random noise ensures that the generator can produce a variety of latent vectors z_gen.

The generator learns to **map the noise distribution** to the real latent space distribution p(z_real).

---

## **4. The Big Picture**
To summarize the generator’s learning process:
1. The generator \( G \) starts with random weights and outputs garbage latent vectors z_gen.
2. The discriminator \( D \) learns to classify real latent vectors z_real and fake ones z_gen.
3. The generator receives gradients from the discriminator, telling it how to adjust its weights to produce better latent vectors.
4. Over time, the generator learns to transform random noise into realistic latent vectors z_gen that match the distribution of z_real.

Once training is complete:
- The generator can take random noise as input and output new latent vectors z_gen.
- These generated latent vectors can be passed to the **autoencoder’s decoder** to produce new CAD sequences.


# PointNet

To train **PointNet++** to project a point cloud into the same latent space as your CAD model autoencoder, the goal is to ensure that the **PointNet++ encoder** produces latent vectors that align with the latent vectors \( c \) generated by the **pre-trained autoencoder**. Here's a step-by-step process:

---

### 1. **Overview of the Problem**
- You already have a pre-trained autoencoder that maps CAD command sequences to a **256-dimensional latent space** \( c \).
- The PointNet++ encoder will process **point clouds** (e.g., 3D scanned data) and map them to the **same latent space** \( c \).
- To achieve this, you need to train PointNet++ so that its latent vectors **match the latent vectors** produced by the autoencoder.

---

### 2. **Data Preparation**
1. **Input Data**: For each CAD model \( M \), you need:
   - Its **point cloud representation** \( P \) (e.g., sampled points from the 3D surface of the CAD model).
   - Its corresponding **latent vector \( c \)** produced by the autoencoder (already trained).

   This gives you paired data:  
   \[
   (P, c)
   \]  
   where \( P \) is the point cloud, and \( c \) is the target latent vector.

2. **Point Cloud Sampling**:
   - Ensure the point clouds are consistent (e.g., uniformly sampled points from the CAD model surface).
   - Use a fixed number of points (e.g., 1024 or 2048 points) for all point clouds to standardize input to PointNet++.

---

### 3. **Architecture Setup**
1. **Pre-trained Autoencoder**:
   - The autoencoder is fixed (frozen weights). It acts as a **reference** for the target latent space \( c \).

2. **PointNet++ Encoder**:
   - Use the PointNet++ architecture to process the point cloud \( P \) and output a latent vector \( \hat{c} \) of size 256 (matching the autoencoder's latent space).

   Architecture:
   \[
   P \rightarrow \text{PointNet++ Encoder} \rightarrow \hat{c}.
   \]

3. **Loss Function**:
   - Use a **regression loss** to minimize the difference between the latent vector \( \hat{c} \) produced by PointNet++ and the latent vector \( c \) from the autoencoder:
     \[
     L = \| c - \hat{c} \|^2
     \]
   - This ensures that the PointNet++ encoder learns to produce latent vectors that align with the autoencoder's latent space.
